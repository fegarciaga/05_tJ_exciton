# Bethe-Salpeter estática para un ángulo fijo

Este notebook calcula el propagador excitónico retarded

$$D^R_{\mu\nu}(\Omega;\theta)=\left[V^{-1}\delta_{\mu\nu}-\Pi^R_{\mu\nu}(\Omega;\theta)\right]^{-1},$$

para dos capas electrónicas acopladas a magnetizaciones clásicas $\mathbf M_1$ y $\mathbf M_2$ con ángulo relativo fijo $\theta$.

El punto importante es que seguimos la estructura del documento: partimos de la burbuja en el contorno, tomamos su componente retardada con reglas de Langreth, y recién después pasamos a frecuencia.


## Espacios: capa, spin y vértice excitónico

La base microscópica completa es

$$\Psi=(c^1_\uparrow,c^1_\downarrow,c^2_\uparrow,c^2_\downarrow)^T,$$

que vive en

$$\mathcal H_{\rm layer}\otimes\mathcal H_{\rm spin}.$$

Las matrices de Pauli $\sigma^a_{\rm spin}$ del acoplamiento $s$-$d$ actúan en el **spin físico dentro de cada capa**:

$$h_l(k)=\epsilon_l(k)\sigma^0_{\rm spin}-J_{sd}\,\mathbf M_l\cdot\boldsymbol\sigma_{\rm spin}.$$

En cambio, el campo excitónico

$$B^{\sigma\sigma'}=c^{1\dagger}_\sigma c^2_{\sigma'}$$

conecta capa 2 con capa 1. Su descomposición

$$B^\mu={\rm Tr}_{\rm spin}[\Gamma^\mu B],\qquad \Gamma^\mu=\sigma^\mu_{\rm spin}/\sqrt{2}$$

usa $\Gamma^\mu$ como **vértice en el espacio de spin del par electrón-hueco**. La estructura de capa está en que el vértice conecta $2\to1$; en este notebook trabajamos por bloques, con $G_1$ y $G_2$ como Green functions de spin $2\times2$ de cada capa.

La burbuja del documento se escribe entonces como traza solo sobre spin:

$$\Pi_{\mu\nu}(z,z')=-\frac{i}{L}\sum_k {\rm Tr}_{\rm spin}\left[\Gamma_\mu^\dagger G^1_M(k;z,z')\Gamma_\nu G^2_M(k;z',z)\right].$$


## Componente retardada

Aplicando Langreth y usando invariancia temporal de traslación para una configuración magnética congelada,

$$\Pi^R_{\mu\nu}(\Omega)=-\frac{i}{L}\sum_k\int\frac{d\omega}{2\pi}{\rm Tr}_{\rm spin}\left[\Gamma_\mu^\dagger G^{1,R}_M(k,\omega+\Omega)\Gamma_\nu G^{2,<}_M(k,\omega)+\Gamma_\mu^\dagger G^{1,<}_M(k,\omega+\Omega)\Gamma_\nu G^{2,A}_M(k,\omega)\right].$$

Usamos

$$G_l^R(k,\omega)=[\omega+i\eta_e-h_l(k)]^{-1},\qquad G_l^A=(G_l^R)^\dagger,$$

$$G_l^<(k,\omega)=f(\omega)\left[G_l^A(k,\omega)-G_l^R(k,\omega)\right].$$

La suma de bandas se deja como benchmark, no como cálculo principal.


```julia
using LinearAlgebra
using Printf
using PyPlot

rc("font", family="serif")
rc("font", size=13)
rc("axes", labelsize=14, titlesize=14)
rc("legend", fontsize=10)
```

```julia
# -----------------------------
# Parámetros físicos y numéricos
# Los defaults son de notebook/smoke test. Para producción subir L, NΩ y Nω.
# -----------------------------
L = 100#36                        # puntos de momento; subir para producción
γ = 1.0                       # hopping intralayer
Δ_layer = 8.0                 # offset energético entre capas: ε1=+Δ/2+εk, ε2=-Δ/2+εk
Jsd = 0.35                    # acoplamiento s-d
Mmag = 1.0                    # magnitud de las magnetizaciones clásicas
θ = 0.0#π/4                       # ángulo fijo entre M1 y M2
V = -1.2                      # atracción excitónica efectiva en D=(V^{-1}-Π)^{-1}
T = 0.08                      # temperatura electrónica
μchem = 0.0                   # potencial químico
ηe = 0.08                     # ensanchamiento de los Green functions electrónicos
ηΩ = 2ηe                      # anchura efectiva del benchmark de bandas

Ωmin, Ωmax, NΩ = 0.0, 15.0, 401#121 -20.0
ωmin, ωmax, Nω = -15.0, 15.0, 601#601#401
ks = collect(range(-π, stop=π - 2π/L, length=L))
Ωs = collect(range(Ωmin, Ωmax, length=NΩ))
ωs = collect(range(ωmin, ωmax, length=Nω))
dω = ωs[2] - ωs[1];
```

```julia
# Pauli en el espacio de spin físico.
σ_spin_0 = ComplexF64[1 0; 0 1]
σ_spin_x = ComplexF64[0 1; 1 0]
σ_spin_y = ComplexF64[0 -im; im 0]
σ_spin_z = ComplexF64[1 0; 0 -1]
σ_spin = [σ_spin_0, σ_spin_x, σ_spin_y, σ_spin_z]

# Vértices excitónicos. Son matrices 2x2 entre las copias de spin de capa 2 y capa 1.
# La estructura inter-capa no está en Γ; está en el bilineal c1† Γ c2.
Γ_exciton = σ_spin ./ sqrt(2)
channel_labels = ["0", "x", "y", "z"];
```

```julia
ϵk(k; γ=γ) = -2γ*cos(k)

function fermi(E; T=T, μ=μchem)
    x = (E - μ) / T
    x > 60 && return 0.0
    x < -60 && return 1.0
    return 1 / (exp(x) + 1)
end

function h_layer(k, M; layer::Int)
    offset = layer == 1 ? +Δ_layer/2 : Δ_layer/2
    energy = (ϵk(k) + offset) * σ_spin_0 - Jsd * (M[1]*σ_spin_x + M[2]*σ_spin_y + M[3]*σ_spin_z)
    energy .*= (-1)^(layer+1)
    return energy
end

function magnetizations(θ)
    M1 = Mmag .* [0.0, 0.0, 1.0]
    M2 = Mmag .* [sin(θ), 0.0, cos(θ)]
    return M1, M2
end

function diagonalize_layers(ks, θ)
    M1, M2 = magnetizations(θ)

    E1 = zeros(Float64, 2, length(ks))
    E2 = zeros(Float64, 2, length(ks))
    U1 = Vector{Matrix{ComplexF64}}(undef, length(ks))
    U2 = Vector{Matrix{ComplexF64}}(undef, length(ks))

    for (ik, k) in enumerate(ks)
        F1 = eigen(Hermitian(h_layer(k, M1; layer=1)))
        F2 = eigen(Hermitian(h_layer(k, M2; layer=2)))
        E1[:, ik] .= F1.values
        E2[:, ik] .= F2.values
        U1[ik] = Matrix{ComplexF64}(F1.vectors)
        U2[ik] = Matrix{ComplexF64}(F2.vectors)
    end

    return (; M1, M2, E1, E2, U1, U2)
end

function green_R(h, ω; η=ηe)
    return inv((ω + im*η) * σ_spin_0 - h)
end

green_A(h, ω; η=ηe) = green_R(h, ω; η=η)'

green_less(h, ω; η=ηe) = fermi(ω) * (green_A(h, ω; η=η) - green_R(h, ω; η=η));
```

```julia
#ks = -pi:0.1:pi
e1(k) = eigen(h_layer(k, [0.,0.,1.0]; layer=1)).values
e2(k) = eigen(h_layer(k, [0.5,0.0,0.0]; layer=2)).values
# #eigen(h_layer(k, [0.,0.,1.0]; layer=1))
plt.ylabel(L"E_{1/2}(k)")
plt.xlabel("k/a")
plt.plot(ks,e1.(ks))
plt.plot(ks,e2.(ks))
#(E1,E2, )=diagonalize_layers(ks, 0.0)
```

```julia
function polarization_R_langreth(Ωs, ωs, ks, θ)
    M1, M2 = magnetizations(θ)
    Π = zeros(ComplexF64, 4, 4, length(Ωs))

    for k in ks
        h1 = h_layer(k, M1; layer=1)
        h2 = h_layer(k, M2; layer=2)

        for (iΩ, Ω) in enumerate(Ωs)
            acc = zeros(ComplexF64, 4, 4)
            for ω in ωs
                G1R = green_R(h1, ω + Ω)
                G1L = green_less(h1, ω + Ω)
                G2A = green_A(h2, ω)
                G2L = green_less(h2, ω)

                for μ in 1:4, ν in 1:4
                    Γμ = Γ_exciton[μ]
                    Γν = Γ_exciton[ν]
                    acc[μ, ν] += tr(Γμ' * G1R * Γν * G2L + Γμ' * G1L * Γν * G2A)
                end
            end
            Π[:, :, iΩ] .+= (-im) * acc * dω / (2π * length(ks))
        end
    end

    return Π
end

function polarization_R_band_benchmark(Ωs, ks, θ)
    data = diagonalize_layers(ks, θ)
    Π = zeros(ComplexF64, 4, 4, length(Ωs))

    for (ik, _) in enumerate(ks)
        U1k = data.U1[ik]
        U2k = data.U2[ik]
        Mμ = [U1k' * Γ_exciton[μ] * U2k for μ in 1:4]

        for a in 1:2, b in 1:2
            ΔE = data.E1[a, ik] - data.E2[b, ik]
            occ = fermi(data.E2[b, ik]) - fermi(data.E1[a, ik])

            for (iΩ, Ω) in enumerate(Ωs)
                denom = Ω + im*ηΩ - ΔE
                pref = occ / denom / length(ks)
                for μ in 1:4, ν in 1:4
                    Π[μ, ν, iΩ] += pref * Mμ[μ][a, b] * conj(Mμ[ν][a, b])
                end
            end
        end
    end

    return Π, data
end

function exciton_propagator(Π; V=V)
    D = similar(Π)
    VinvI = (1/V) * Matrix{ComplexF64}(I, 4, 4)
    for iΩ in axes(Π, 3)
        D[:, :, iΩ] .= inv(VinvI - Π[:, :, iΩ])
    end
    return D
end

function spectral_trace(D)
    return [-2 * imag(tr(D[:, :, iΩ])) for iΩ in axes(D, 3)]
end;
```

```julia
# println("Calculando Pi^R con Langreth + integral en frecuencia...")
# Π = polarization_R_langreth(Ωs, ωs, ks, θ)
# D = exciton_propagator(Π)
# AD = spectral_trace(D)

println("Calculando benchmark de bandas...")
Π_band, layer_data = polarization_R_band_benchmark(Ωs, ks, θ)
D_band = exciton_propagator(Π_band)
AD_band = spectral_trace(D_band)

println("M1 = ", layer_data.M1)
println("M2 = ", layer_data.M2)
println("theta/pi = ", θ/π)
println("cos(theta) = ", dot(layer_data.M1, layer_data.M2)/(norm(layer_data.M1)*norm(layer_data.M2)))
```

## Figura 1: continuo electrón-hueco bare

Antes de mirar el propagador excitónico, conviene ver las transiciones de una partícula que forman el continuo bare:

$$\Delta E_{ab}(k)=E^1_a(k)-E^2_b(k).$$

Si aparece un pico fuerte de $D^R$ cerca de un borde o separado de este continuo, lo interpretamos como modo excitónico ligado/colectivo. Esta figura no usa todavía $V$ ni la BSE; solo muestra la cinemática de pares electrón-hueco para el ángulo fijo.


```julia
fig, ax = subplots(figsize=(6.8, 4.4))
colors = ["#4d4d4d", "#737373", "#969696", "#bdbdbd"]
idx = 1
for a in 1:2, b in 1:2
    ΔEab = layer_data.E1[a, :] .- layer_data.E2[b, :]
    ax.plot(ks, ΔEab; color=colors[idx], lw=1.5, label="a=$a, b=$b")
    idx += 1
end
ax.axhline(0, color="black", lw=0.8, alpha=0.45)
ax.set_xlabel("k")
ax.set_ylabel("Delta E_ab(k)")
ax.set_title("Bare electron-hole continuum, theta/pi = $(round(θ/π; digits=3))")
ax.set_xlim(first(ks), last(ks))
ax.grid(alpha=0.18)

ax.legend(frameon=false, ncol=2)
fig.tight_layout()
```

## Figura 2: respuesta bare vs respuesta excitónica BSE

Primero calculamos la respuesta electrón-hueco **bare**

$$A_\Pi(\Omega)=-2\,\mathrm{Im}\,\mathrm{Tr}_\mu \Pi^R_{\mu\mu}(\Omega).$$

Esta curva viene de la burbuja y representa el continuo de pares electrón-hueco antes de resumir la interacción atractiva.

Luego resolvemos la BSE/RPA

$$D^R(\Omega)=[V^{-1}I-\Pi^R(\Omega)]^{-1}$$

y graficamos

$$A_D(\Omega)=-2\,\mathrm{Im}\,\mathrm{Tr}_\mu D^R_{\mu\mu}(\Omega).$$

La comparación $A_\Pi$ vs $A_D$ es la que muestra qué hace la interacción excitónica: si $D^R$ desarrolla un pico fuerte, desplazado o separado del continuo bare, eso indica un modo excitónico o una resonancia excitónica. También marcamos el mínimo de $|\det[V^{-1}I-\Pi^R]|$, que localiza la condición de polo BSE.

## Figura 3: composición en canales carga/spin

Finalmente separamos los canales diagonales del propagador excitónico:

$$A_\mu(\Omega)=-2\,\mathrm{Im}\,D^R_{\mu\mu}(\Omega),\qquad \mu=0,x,y,z.$$

Esto muestra si el modo excitónico es principalmente de carga, spin longitudinal/transversal, o si la no colinealidad entre $\mathbf M_1$ y $\mathbf M_2$ mezcla varios canales.


```julia
fig, ax = subplots(figsize=(6.8, 4.4))
for μ in 1:4
    Aμμ = [-2 * imag(D[μ, μ, iΩ]) for iΩ in axes(D, 3)]
    ax.plot(Ωs, Aμμ; lw=1.8, label=channel_labels[μ])
end
ax.set_xlabel("Omega")
ax.set_ylabel("-2 Im D^R_mu,mu")
ax.set_title("Charge/spin channel content")
ax.grid(alpha=0.18)
ax.legend(frameon=false, title=L"$\mu$")
ax.set_xlim(0, 15)





fig.tight_layout()
```

## Diagnóstico adicional: mezcla entre canales

Como chequeo numérico rápido, imprimimos

$$\max_\Omega |\Pi^R_{\mu\nu}(\Omega)|.$$

Para configuraciones colineales debería haber una estructura más bloque-diagonal. Para ángulos no colineales, la mezcla entre canales es parte de la física esperada.


```julia
mixing = zeros(Float64, 4, 4)
for μ in 1:4, ν in 1:4
    mixing[μ, ν] = maximum(abs.(Π[μ, ν, :]))
end

println("max_Omega |Pi_{μν}(Omega;theta)|")
print("      ")
for label in channel_labels
    @printf("%12s", label)
end
println()
for μ in 1:4
    @printf("%4s  ", channel_labels[μ])
    for ν in 1:4
        @printf("%12.5e", mixing[μ, ν])
    end
    println()
end
```

## Qué mirar después

Para convertir esto en una caracterización por ángulo, el siguiente paso es envolver las celdas anteriores en una función `solve_angle(θ)` y barrer `θ ∈ [0,π]`. La figura central después será un mapa de color

$$A_D(\Omega,\theta)=-2\,\mathrm{Im}\,\mathrm{Tr}\,D^R(\Omega;\theta),$$

con eje horizontal `Ω` y eje vertical `θ` o `cos(θ)`.
