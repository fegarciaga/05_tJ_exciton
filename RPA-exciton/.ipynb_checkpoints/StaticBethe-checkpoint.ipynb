# Bethe-Salpeter estática para un ángulo fijo

Este notebook construye el primer objeto que queremos estudiar: el propagador excitónico retarded

$$D^R_{\mu\nu}(\Omega;\theta)=\left[V^{-1}\delta_{\mu\nu}-\Pi^R_{\mu\nu}(\Omega;\theta)\right]^{-1},$$

para dos capas electrónicas acopladas a magnetizaciones clásicas $\mathbf M_1$ y $\mathbf M_2$ con ángulo relativo fijo $\theta$.

La idea es la analogía directa con el cálculo estacionario del propagador en `KBE-DrivenElectrons`: allí se construía $G^R(k,\omega)$ y su espectro; aquí construimos la burbuja electrón-hueco $\Pi^R$ y el propagador colectivo excitónico $D^R$.

En este primer notebook no hacemos dinámica en tiempo ni autoconsistencia con $M(t)$. Fijamos una configuración magnética congelada y calculamos el espectro excitónico correspondiente.

## Modelo y componente retardada

Usamos dos capas de electrones con spin. Para cada capa,

$$h_l(k)=\epsilon_l(k)\sigma_0 - J_{sd}\,\mathbf M_l\cdot\boldsymbol\sigma.$$

Por simetría global de spin, si no hay spin-orbit ni anisotropías explícitas, podemos escoger

$$\mathbf M_1=M\hat z,$$

$$\mathbf M_2=M(\sin\theta\,\hat x + \cos\theta\,\hat z).$$

Entonces el problema debe depender solamente de $\theta$ o, equivalentemente, de $\cos\theta=\hat M_1\cdot\hat M_2$.

El operador excitónico local se descompone en canales carga/spin

$$B^\mu = c_1^\dagger \Gamma^\mu c_2,\qquad \Gamma^\mu=\sigma^\mu/\sqrt{2},\qquad \mu=0,x,y,z.$$

En el documento, la burbuja en el contorno es

$$\Pi_{\mu\nu}(z,z')=-\frac{i}{L}\sum_k \mathrm{Tr}\left[\Gamma_\mu^\dagger G^1_M(k;z,z')\Gamma_\nu G^2_M(k;z',z)\right].$$

Aquí calculamos la componente retardada aplicando Langreth y luego Fourier. En equilibrio y para magnetizaciones congeladas,

$$\Pi^R_{\mu\nu}(\Omega)=-\frac{i}{L}\sum_k\int\frac{d\omega}{2\pi}\mathrm{Tr}\left[\Gamma_\mu^\dagger G^{1,R}_M(k,\omega+\Omega)\Gamma_\nu G^{2,<}_M(k,\omega)+\Gamma_\mu^\dagger G^{1,<}_M(k,\omega+\Omega)\Gamma_\nu G^{2,A}_M(k,\omega)\right].$$

Usamos

$$G_l^R(k,\omega)=[\omega+i\eta_e-h_l(k)]^{-1},\qquad G_l^A=(G_l^R)^\dagger,$$

$$G_l^<(k,\omega)=f(\omega)\left[G_l^A(k,\omega)-G_l^R(k,\omega)\right].$$

Luego resolvemos la BSE/RPA

$$D^R_{\mu\nu}(\Omega;\theta)=\left[V^{-1}\delta_{\mu\nu}-\Pi^R_{\mu\nu}(\Omega;\theta)\right]^{-1}.$$

La fórmula de bandas queda como chequeo: para electrones no interactuantes, la integral anterior debe reducirse a la suma de transiciones entre autestados, salvo diferencias numéricas por ensanchamiento, ventana finita de integración y la anchura efectiva usada para representar la convolución de dos Green functions.


```julia
using LinearAlgebra
using Printf
```

```julia
# -----------------------------
# Parámetros físicos y numéricos
# Los defaults son de notebook/smoke test. Para producción subir L, NΩ y Nω.
# -----------------------------
L = 36                        # puntos de momento; subir para producción
γ = 1.0                       # hopping intralayer
Δ_layer = 1.5                 # offset energético entre capas: ε1=+Δ/2+εk, ε2=-Δ/2+εk
Jsd = 0.35                    # acoplamiento s-d
Mmag = 1.0                    # magnitud de las magnetizaciones clásicas
θ = π/2                       # ángulo fijo entre M1 y M2
V = 1.2                       # atracción excitónica efectiva en D=(V^{-1}-Π)^{-1}
T = 0.08                      # temperatura electrónica
μchem = 0.0                   # potencial químico
ηe = 0.08                     # ensanchamiento de los Green functions electrónicos
ηΩ = 2ηe                      # anchura efectiva del benchmark de bandas

Ωmin, Ωmax, NΩ = -4.0, 4.0, 121
ωmin, ωmax, Nω = -7.0, 7.0, 401
ks = collect(range(-π, stop=π - 2π/L, length=L))
Ωs = collect(range(Ωmin, Ωmax, length=NΩ))
ωs = collect(range(ωmin, ωmax, length=Nω))
dω = ωs[2] - ωs[1];
```

```julia
# Matrices de Pauli y vértices Γ^μ = σ^μ/sqrt(2)
σ0 = ComplexF64[1 0; 0 1]
σx = ComplexF64[0 1; 1 0]
σy = ComplexF64[0 -im; im 0]
σz = ComplexF64[1 0; 0 -1]

σs = [σ0, σx, σy, σz]
Γs = σs ./ sqrt(2)
channel_labels = ["0", "x", "y", "z"];
```

```julia
ϵk(k; γ=γ) = -2γ*cos(k)

function fermi(E; T=T, μ=μchem)
    x = (E - μ) / T
    x > 60 && return 0.0
    x < -60 && return 1.0
    return 1 / (exp(x) + 1)
end

function h_layer(k, M; layer::Int)
    offset = layer == 1 ? +Δ_layer/2 : -Δ_layer/2
    return (ϵk(k) + offset) * σ0 - Jsd * (M[1]*σx + M[2]*σy + M[3]*σz)
end

function magnetizations(θ)
    M1 = Mmag .* [0.0, 0.0, 1.0]
    M2 = Mmag .* [sin(θ), 0.0, cos(θ)]
    return M1, M2
end

function diagonalize_layers(ks, θ)
    M1, M2 = magnetizations(θ)

    E1 = zeros(Float64, 2, length(ks))
    E2 = zeros(Float64, 2, length(ks))
    U1 = Vector{Matrix{ComplexF64}}(undef, length(ks))
    U2 = Vector{Matrix{ComplexF64}}(undef, length(ks))

    for (ik, k) in enumerate(ks)
        F1 = eigen(Hermitian(h_layer(k, M1; layer=1)))
        F2 = eigen(Hermitian(h_layer(k, M2; layer=2)))
        E1[:, ik] .= F1.values
        E2[:, ik] .= F2.values
        U1[ik] = Matrix{ComplexF64}(F1.vectors)
        U2[ik] = Matrix{ComplexF64}(F2.vectors)
    end

    return (; M1, M2, E1, E2, U1, U2)
end

function green_R(h, ω; η=ηe)
    return inv((ω + im*η) * σ0 - h)
end

green_A(h, ω; η=ηe) = green_R(h, ω; η=η)'

green_less(h, ω; η=ηe) = fermi(ω) * (green_A(h, ω; η=η) - green_R(h, ω; η=η));
```

```julia
function polarization_R_langreth(Ωs, ωs, ks, θ)
    M1, M2 = magnetizations(θ)
    Π = zeros(ComplexF64, 4, 4, length(Ωs))

    for k in ks
        h1 = h_layer(k, M1; layer=1)
        h2 = h_layer(k, M2; layer=2)

        for (iΩ, Ω) in enumerate(Ωs)
            acc = zeros(ComplexF64, 4, 4)
            for ω in ωs
                G1R = green_R(h1, ω + Ω)
                G1L = green_less(h1, ω + Ω)
                G2A = green_A(h2, ω)
                G2L = green_less(h2, ω)

                for μ in 1:4, ν in 1:4
                    acc[μ, ν] += tr(Γs[μ]' * G1R * Γs[ν] * G2L + Γs[μ]' * G1L * Γs[ν] * G2A)
                end
            end
            Π[:, :, iΩ] .+= (-im) * acc * dω / (2π * length(ks))
        end
    end

    return Π
end

function polarization_R_band_benchmark(Ωs, ks, θ)
    data = diagonalize_layers(ks, θ)
    Π = zeros(ComplexF64, 4, 4, length(Ωs))

    for (ik, _) in enumerate(ks)
        U1k = data.U1[ik]
        U2k = data.U2[ik]
        Mμ = [U1k' * Γs[μ] * U2k for μ in 1:4]

        for a in 1:2, b in 1:2
            ΔE = data.E1[a, ik] - data.E2[b, ik]
            occ = fermi(data.E2[b, ik]) - fermi(data.E1[a, ik])

            for (iΩ, Ω) in enumerate(Ωs)
                denom = Ω + im*ηΩ - ΔE
                pref = occ / denom / length(ks)
                for μ in 1:4, ν in 1:4
                    Π[μ, ν, iΩ] += pref * Mμ[μ][a, b] * conj(Mμ[ν][a, b])
                end
            end
        end
    end

    return Π, data
end

function exciton_propagator(Π; V=V)
    D = similar(Π)
    VinvI = (1/V) * Matrix{ComplexF64}(I, 4, 4)
    for iΩ in axes(Π, 3)
        D[:, :, iΩ] .= inv(VinvI - Π[:, :, iΩ])
    end
    return D
end

function spectral_trace(D)
    return [-2 * imag(tr(D[:, :, iΩ])) for iΩ in axes(D, 3)]
end;
```

```julia
println("Calculando Pi^R con Langreth + integral en frecuencia...")
Π = polarization_R_langreth(Ωs, ωs, ks, θ)
D = exciton_propagator(Π)
AD = spectral_trace(D)

println("Calculando benchmark de bandas...")
Π_band, layer_data = polarization_R_band_benchmark(Ωs, ks, θ)
D_band = exciton_propagator(Π_band)
AD_band = spectral_trace(D_band)

println("M1 = ", layer_data.M1)
println("M2 = ", layer_data.M2)
println("theta/pi = ", θ/π)
println("cos(theta) = ", dot(layer_data.M1, layer_data.M2)/(norm(layer_data.M1)*norm(layer_data.M2)))
```

```julia
# Diagnóstico de las bandas polarizadas. Este notebook evita dependencias
# externas de plotting para que corra en el entorno actual.
println("Bandas polarizadas por magnetizaciones congeladas")
for layer in 1:2, band in 1:2
    E = layer == 1 ? layer_data.E1[band, :] : layer_data.E2[band, :]
    @printf("layer %d band %d: min(E) = %+8.5f, max(E) = %+8.5f\n", layer, band, minimum(E), maximum(E))
end

println("\nPrimeros puntos de k y energías:")
for ik in 1:5
    @printf("k=%+7.4f | E1=(%+7.4f,%+7.4f) | E2=(%+7.4f,%+7.4f)\n",
        ks[ik], layer_data.E1[1,ik], layer_data.E1[2,ik], layer_data.E2[1,ik], layer_data.E2[2,ik])
end
```

```julia
# Diagnóstico espectral: pico principal de Tr D y picos por canal.
i_peak = argmax(AD)
i_peak_band = argmax(AD_band)
@printf("Langreth: pico principal -2 Im Tr D^R: Omega = %+8.5f, peso = %+10.6f\n", Ωs[i_peak], AD[i_peak])
@printf("Band benchmark: pico principal -2 Im Tr D^R: Omega = %+8.5f, peso = %+10.6f\n", Ωs[i_peak_band], AD_band[i_peak_band])

relerrΠ = maximum(abs.(Π .- Π_band)) / max(maximum(abs.(Π_band)), eps())
@printf("Diferencia relativa max |Pi_Langreth - Pi_band| / max|Pi_band| = %.6e\n", relerrΠ)

println("\nPicos por canal diagonal D_{μμ} usando Langreth:")
for μ in 1:4
    Aμμ = [-2 * imag(D[μ, μ, iΩ]) for iΩ in axes(D, 3)]
    ip = argmax(Aμμ)
    @printf("channel %-1s: Omega_peak = %+8.5f, weight = %+10.6f\n", channel_labels[μ], Ωs[ip], Aμμ[ip])
end

detvals = [abs(det((1/V) * Matrix{ComplexF64}(I, 4, 4) - Π[:, :, iΩ])) for iΩ in axes(Π, 3)]
i_det = argmin(detvals)
@printf("\nMínimo de |det(V^{-1}I - Pi)|: Omega = %+8.5f, value = %.6e\n", Ωs[i_det], detvals[i_det])
```

```julia
# Mezcla entre canales: una forma simple de ver si la BSE queda diagonal o no.
mixing = zeros(Float64, 4, 4)
for μ in 1:4, ν in 1:4
    mixing[μ, ν] = maximum(abs.(Π[μ, ν, :]))
end

println("max_Omega |Pi_{μν}(Omega;theta)|")
print("      ")
for label in channel_labels
    @printf("%12s", label)
end
println()
for μ in 1:4
    @printf("%4s  ", channel_labels[μ])
    for ν in 1:4
        @printf("%12.5e", mixing[μ, ν])
    end
    println()
end
```
